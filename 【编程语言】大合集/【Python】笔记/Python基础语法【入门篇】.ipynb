## 三、数据类型

### 1. 数值类型

#### 整数类型

整数可以用不同进制表示：

- 十进制：直接书写，如 `987`
- 二进制：以 `0b` 开头，如 `0b101010`
- 八进制：以 `0o` 开头，如 `0o765`

````python
num =987
num2=0b101010   #2进制表示
num3=0o765      #8进制表示
num4=0x87abf    #16进制表示

print(num)
print(num2)
print(num3)
print(num4)

'''
输出的是他们自动以十进制输出
——————-输出如下———————
987
42
501
555711
——————————————————
'''
````

#### 浮点数类型

浮点数用于表示小数，也可以使用科学计数法，例如 `1.99E143`。


````python
x=10  
y=10.0  
z=1.99E143  #科学计数法  
print("x的数据类型：",type(x))  
print("y的数据类型：",type(y))  
print("z的数据类型：",type(z))

'''
——————-输出如下———————
x的数据类型： <class 'int'>
y的数据类型： <class 'float'>
z的数据类型： <class 'float'>
——————————————————
'''
````

#### 复数类型

复数由实部和虚部组成，虚部以 `j` 结尾，例如 `123 + 456j`。可以用 `.real` 取实部、`.imag` 取虚部。


````python
'''
复数类型
实数用 `.real` ，虚数用`.imag` 表示
'''

x=123+456j
print("实数部分：",x.real)
print("虚数部分：",x.imag)


'''——————-输出如下———————
实数部分： 123.0
虚数部分： 456.0
——————————————————'''
````

````python

li = [ 'runoob', 786 , 2.23, 'john', 70.2 ]
tinylist = [123, 'john'] 
print(li) # 输出完整列表 
print(li[0])# 输出列表的第一个元素
print(li[1:3]) # 输出第二个至第三个元素 
print(li[2:]) # 输出从第三个开始至列表末尾的所有元素 
print(tinylist * 2) # 输出列表两次 
print(li + tinylist) # 打印组合的列表


'''——————-输出如下———————
['runoob', 786, 2.23, 'john', 70.2]
runoob
[786, 2.23]
[2.23, 'john', 70.2]
[123, 'john', 123, 'john']
['runoob', 786, 2.23, 'john', 70.2, 123, 'john']
——————————————————'''
````

### 3. 类型转换

#### 显式类型转换

简单转换和字符串转换：

- `int(1231.11)` -> `1231`：浮点数转整数，会截断小数部分
- `float('13.5')` -> `13.5`：数字字符串转浮点数
- `int('ad')` -> 抛出 `ValueError`：非法字符串无法转换为整数

序列之间的转换：

- `set([1, 2, 3])`：将列表转换为集合
- `tuple({5, 6, 7})`：将集合转换为元组
- `list('hello')`：将字符串转换为列表
- `dict(zip([1, 2], [3, 4]))`：将两个序列组合成字典 `{1: 3, 2: 4}`


````python
# 简单转换
b = int(1231.11)
print(b)  # 1231

# 字符串的转换
c = float('13.5')
print(c)  # 13.5

# 序列间的转换
print(set([1, 2, 3]))              # {1, 2, 3}
print(tuple({5, 6, 7}))            # 输出顺序可能不同
print(list('hello'))               # ['h', 'e', 'l', 'l', 'o']
print(dict(zip([1, 2], [3, 4])))   # {1: 3, 2: 4}
````

## 四、语法结构

### 分支结构

分支结构用于根据条件选择不同代码路径。Python 中 `if` 后面要写冒号，并用缩进表示属于该分支的代码块。


````python
#python中的if后面是冒号，且严格缩进，中间的分支用elif
num=5
if num>0:
	print("是正数")
elif num<0:
	print("是负数")
else:
	print("是零")
````

`if` 支持嵌套，但嵌套层级越多越难读；写代码时要特别注意缩进是否属于预期分支。

### 循环结构

`for` 循环常用于遍历序列，例如 `list`、`tuple`、`str`。


````python
numbers=[6,5,3,5,131,6]
sum=0;

for val in numbers:
	sum=sum+val

print("总和是",sum)
````

`range(start, stop, step)` 用于生成整数序列：从 `start` 开始，到 `stop` 之前结束，不包含 `stop`，每次增加 `step`。


````python
genre = ['pop', 'rock', 'jazz']

for i in range(len(genre)):
    print("I like", genre[i])
````

### 循环中的 `else`

`for` 循环可以带一个可选的 `else` 代码块：如果循环正常结束，`else` 会执行；如果循环中途遇到 `break`，`else` 会被跳过。

### `break`、`continue` 与 `pass`

- `break`：立刻结束整个循环。
- `continue`：跳过本轮循环剩余语句，进入下一轮。
- `pass`：占位语句，不做任何事情，常用于暂时保留代码结构。

### 面向对象技术简介

- **类（Class）**：描述具有相同属性和方法的一类对象。对象是类的实例。
- **对象（Object）**：由类创建出来的具体个体，包含数据和行为。
- **方法**：定义在类中的函数，通常通过对象调用。
- **类变量**：定义在类中、方法外，对所有实例共享。
- **实例变量**：绑定在具体对象上的变量，通常写作 `self.xxx`。
- **数据成员**：类变量或实例变量，用于保存对象相关数据。
- **继承**：子类复用父类已有属性和方法，表达“是一种”的关系。
- **方法重写**：子类重新定义父类已有方法，以满足自己的需求。
- **实例化**：根据类创建对象的过程。

### 类的方法


````python
#!/usr/bin/python3
class MyClass:
    """一个简单的类实例"""
    i = 12345

    def f(self):
        return 'hello world'

# 实例化类
x = MyClass()  # 访问类的属性和方法
print("MyClass 类的属性 i 为：", x.i)
print("MyClass 类的方法 f 输出为：", x.f())
````

#### `__init__()` 方法

`__init__()` 是初始化方法。创建对象时，Python 会自动调用它，用来给实例变量赋初始值。


````python
#!/usr/bin/python3

class Complex:
    def __init__(self, realpart, imagpart):
        self.r = realpart
        self.i = imagpart
x = Complex(3.0, -4.5)
print(x.r, x.i)   # 输出结果：3.0 -4.5
````

#### 普通方法

类的方法和普通函数很像，但实例方法必须有一个额外的第一个参数，按惯例命名为 `self`。`self` 表示当前对象本身。


````python
class MyClass:
    def __init__(self, value):
        self.value = value

    def display_value(self):
        print(self.value)

obj = MyClass(42) 

obj.display_value() # 输出 42
````

### 类的继承

子类（派生类）会继承父类（基类）的属性和方法。继承能复用已有代码，也能通过方法重写改变父类行为。

下面两段代码分别演示：

- 子类自动使用父类方法。
- 子类重写父类方法。


````python
class Fruit:
    color = "绿色"  # 类变量

    def harvest(self, color):
        print("水果是：" + color + "的！")
        print("水果已经收获......")
        print("水果原来是：" + Fruit.color + "的！")


class Apple(Fruit):
    color = "红色"  # 覆盖父类的类变量

    def __init__(self):
        print("\n我是苹果")
        # 自动继承父类的 harvest 方法


class Orange(Fruit):
    color = "橙色"

    def __init__(self):
        print("\n我是橘子")


apple = Apple()  # 创建实例时调用 __init__ 方法
apple.harvest(apple.color)  # 调用继承来的 harvest 方法

orange = Orange()
orange.harvest(orange.color)
````

### 运算符重载

Python 的运算符（如 `+`、`-`、`==`）在底层都对应一个**魔法方法**（双下划线方法）。例如 `a + b` 实际上调用的是 `a.__add__(b)`，两者完全等价。

在自定义类中重写这些魔法方法，就能改变运算符的行为，这称为**运算符重载**。例如让两个矩阵对象用 `+` 实现“对应元素相加”，而不是像列表那样直接拼接。

常用运算符与魔法方法的对应关系：

| 运算符 | 魔法方法 | 说明 |
|--------|----------|------|
| `+` | `__add__` | 加法 |
| `-` | `__sub__` | 减法 |
| `*` | `__mul__` | 乘法 |
| `==` | `__eq__` | 相等比较 |
| `<` | `__lt__` | 小于比较 |
| `len()` | `__len__` | 取长度 |
| `print()` | `__repr__` / `__str__` | 控制显示内容 |

````python
# 运算符重载：让 Matrix 支持 + 运算（对应元素相加）
class Matrix:
    def __init__(self, data):
        self.data = data

    def __add__(self, other):           # 重载 + ：a + b 等价于 a.__add__(b)
        result = [[self.data[i][j] + other.data[i][j]
                   for j in range(len(self.data[0]))]
                  for i in range(len(self.data))]
        return Matrix(result)           # 返回一个新的 Matrix

    def __repr__(self):                 # 控制 print 时的显示内容
        return str(self.data)

a = Matrix([[1, 2, 3], [4, 5, 6]])
b = Matrix([[7, 8, 9], [10, 11, 12]])

print(a + b)            # [[8, 10, 12], [14, 16, 18]]
print(a.__add__(b))     # 与 a + b 完全等价

'''——————-输出如下———————
[[8, 10, 12], [14, 16, 18]]
[[8, 10, 12], [14, 16, 18]]
——————————————————'''
````

## 六、模块与包

Python 中的模块（Module）就是一个 `.py` 文件，文件名就是模块名。

模块可以包含函数、类、变量以及可执行语句。通过模块，我们可以把代码拆成可复用的单元，便于管理和维护。

### 模块的作用

- **代码复用**：把常用功能封装到模块中，在多个程序中重复使用。
- **命名空间管理**：不同模块中的同名函数或变量不会直接冲突。
- **代码组织**：按功能拆分代码，使程序结构更清晰。

### 使用 `import` 导入模块

假设同级目录下有两个文件：`test.py` 和 `demo.py`。

`test.py` 中代码如下：


````python
def my_printf():  
    print("Hello World")
````

`demo.py` 中代码如下：


````python
import test  
test.my_printf()
````

如果不想每次都写完整模块名，可以给模块起别名。


````python
import test as m
m.my_printf()
````

也可以使用 `from ... import ...` 直接导入模块中的某个函数。


````python
from test import my_printf
my_printf()
````

如果使用 `from 模块名 import *`，会导入该模块中所有不以下划线 `_` 开头的名字。实际项目中不推荐频繁使用这种写法，因为它容易让命名来源变得不清楚。

### 包是什么

![](https://lin01-image-1373317342.cos.ap-beijing.myqcloud.com/c4cb0fcbefb3fb53b2331e0235e931d4.png)

> **图片说明**：包的结构示例：项目根目录下 `main.py` 作为入口，`chap2` 文件夹中包含 `__init__.py` 和 `test.py`，共同组成一个包。

包可以理解为“装模块的文件夹”。一个包目录中通常包含 `__init__.py`，它用于标记这个目录是一个 Python 包，也可以放包初始化时需要执行的代码。

常见导入方式：

- `from 包名.模块名 import 函数`
- `from 包名.模块名 import *`：导入模块中所有不以下划线 `_` 开头的名字

如果模块里既有函数定义，又有测试代码，而我们只想在直接运行该文件时执行测试代码，可以使用 `if __name__ == '__main__':`。


````python
def my_printf():  
    print("Hello World")  
      
if __name__ == '__main__':  
    print("该模块运行无误")
````

当单独运行该模块时，会输出 `该模块运行无误`。

当该模块被其他文件导入时，`if __name__ == '__main__':` 下面的代码不会执行，但仍然可以调用 `my_printf()`。

### 导入标准模块

标准模块可以直接用 `import 标准模块名` 导入，例如 `import math`、`import random`。

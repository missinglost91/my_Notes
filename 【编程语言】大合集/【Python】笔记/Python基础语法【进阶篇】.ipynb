### 2.1 常见运行时错误

| 错误类型 | 说明 | 示例 |
|---------|------|------|
| `NameError` | 使用了未定义的变量 | `print(x)`  # x 没定义 |
| `TypeError` | 类型操作错误 | `"hello" + 1` |
| `IndexError` | 索引超出范围 | `lst = [1,2]; lst[5]` |
| `KeyError` | 字典键不存在 | `d = {}; d['key']` |
| `ValueError` | 值错误 | `int("abc")` |
| `ZeroDivisionError` | 除以零 | `1 / 0` |
| `FileNotFoundError` | 文件不存在 | `open("no_file.txt")` |
| `AttributeError` | 属性不存在 | `None.some_method()` |

````python
try:                            # 只保护可能出错的代码
    num = 0                       # 故意设为 0 触发 ZeroDivisionError（想试 ValueError 可改成 int('abc')）
    result = 10 / num

except ValueError as e:           # 捕获值错误（输入的不是数字）
    print(f"错误类型：{e}")
    print("输入的不是数字!")

except ZeroDivisionError as e:     # 捕获除零错误
    print(f"错误类型：{e}")
    print("不能除以零！")

else:                             # 没有异常时才执行
    print(f"结果是：{result}")

finally:                          # 无论是否异常都执行（常用于清理资源）
    print("程序结束")
````

## 3. 迭代器与可迭代对象

`range(1, 10)` 会隐式生成一个可迭代对象，其背后正是迭代器机制。


### 3.1 可迭代对象 (Iterable)

- **定义**：能被 for 循环遍历的对象
- **判断方法**：实现了 `__iter__()` 方法 或 `__getitem__()` 方法
- **常见例子**：`list`、`tuple`、`dict`、`set`、`str`

````python
from collections.abc import Iterable

lst = [1, 2, 3]
it = iter(lst)

# 可以被 for 循环
print("--------迭代对象第一次遍历---------")
for i in lst:
    print(i)

print("--------迭代对象第二次遍历---------")
for i in lst:
    print(i)

print("--------迭代器第一次遍历---------")
for i in it:
    print(i)

print("--------迭代器第二次遍历---------")
for i in it:
    print(i)  # 不支持第二遍，迭代器已耗尽，没有元素了

print("----------迭代器取 next---------")
it = iter(lst)
print(next(it))  # 1
print(next(it))  # 2
print(next(it))  # 3
````

### 3.2 for 循环的背后机制

`for x in sequence_iterator` 背后其实是：

```python
it = iter(sequence_iterator)   # 调 __iter__，拿到 self
while True:
    try:
        x = next(it)         # 调 __next__
    except StopIteration:
        break
```


### 3.3 数列迭代器（Progression）

下面用迭代器协议实现一个通用的「数列」框架 `Progression`，再通过继承 + 方法重写派生出等差、等比、斐波那契数列。它综合演示了 `__iter__` / `__next__` 与面向对象的继承。


````python
class Progression:
    def __init__(self, start=0):
        self._current = start

    def __next__(self):
        answer = self._current
        self._advance()              # 调“前进一格”的方法
        return answer

    def __iter__(self):
        return self
    
    def print_progression(self, n):  # 打印 n 项
        for j in range(n):
            print(next(self), end=' ')
        print()
        
class ArithmeticProgression(Progression):
    def __init__(self, increment=1, start=0):
        super().__init__(start)
        self._increment = increment
    def _advance(self):
        self._current += self._increment     # 等差：+increment

class GeometricProgression(Progression):
    def __init__(self, base=2, start=1):
        super().__init__(start)
        self._base = base
    def _advance(self):
        self._current *= self._base          # 等比：×base

class FibonacciProgression(Progression):
    def __init__(self, first=0, second=1):
        super().__init__(first)              # 当前值是第一项
        self._prev = second                  # 额外记住“前一项”
    def _advance(self):
        self._prev, self._current = self._current, self._prev + self._current  # 同时更新

a=ArithmeticProgression(5)  

# 打印等差数列
for i in a:
    print(i)
    if i>=50:
        break

a=ArithmeticProgression(5)  
for i in range(1,10):
        print(next(a))
````

````python
# 改进版：用抽象类（ABC）把 _advance 声明为抽象方法，强制子类必须实现它
from abc import ABCMeta, abstractmethod

class Progression(metaclass=ABCMeta):
    def __init__(self, start=0):
        self._current = start
    
    @abstractmethod              # 抽象方法：必须在子类中实现，否则报错
    def _advance(self):
        self._current += 1           # 默认：每次 +1

    def __next__(self):
        answer = self._current
        self._advance()              # 调“前进一格”的方法
        return answer

    def __iter__(self):
        return self
    
    def print_progression(self, n):  # 打印 n 项
        for j in range(n):
            print(next(self), end=' ')
        print()
        
class ArithmeticProgression(Progression):
    def __init__(self, increment=1, start=0):
        super().__init__(start)
        self._increment = increment
    def _advance(self):
        self._current += self._increment     # 等差：+increment

class GeometricProgression(Progression):
    def __init__(self, base=2, start=1):
        super().__init__(start)
        self._base = base
    def _advance(self):
        self._current *= self._base          # 等比：×base

class FibonacciProgression(Progression):
    def __init__(self, first=0, second=1):
        super().__init__(first)              # 当前值是第一项
        self._prev = second                  # 额外记住“前一项”
    def _advance(self):
        self._prev, self._current = self._current, self._prev + self._current  # 同时更新

a=ArithmeticProgression(5)  


````

### 4.2 序列的打包与解包

#### 打包 (Packing)
把多个值组合成一个序列


````python
# 隐式打包
t = 1, 2, 3
print(t)  # (1, 2, 3)
````

#### 解包 (Unpacking)
把序列拆分成多个独立的值

| 类型 | 语法 | 说明 | 示例 |
|------|------|------|------|
| **基础解包** | `a, b, c = seq` | 变量数必须等于元素数 | `a, b = [1, 2]` |
| **星号解包** | `a, *rest = seq` | `*` 收集剩余元素为列表 | `a, *rest = [1, 2, 3]` |
| **只取部分** | `a, *_ = seq` | `_` 表示丢弃 | `first, *_ = [1, 2, 3]` |
| **函数参数** | `func(*args)` | 列表解包为位置参数 | `add(*[1, 2])` |
| **字典参数** | `func(**kwargs)` | 字典解包为关键字参数 | `func(**{'a': 1})` |

#### 示例代码


````python
# 收集中间部分
first, *middle, last = [1, 2, 3, 4, 5]
print(first)    # 1
print(middle)   # [2, 3, 4]  ← 自动打包成列表
print(last)     # 5
````

````python
def get_user_info():
    return "Alice", 25, "Engineer"

name, age, job = get_user_info()
print(name)  # Alice
print(age)   # 25
print(job)   # Engineer
````

## 5. 函数与文档

### 5.1 docstring 与注释的区别

- **docstring**：写在函数 / 类 / 模块体第一行的字符串，运行时可通过 `__doc__` 和 `help()` 访问，是“文档”。
- **注释**（`#`）：解释实现细节，解释器完全忽略，运行时访问不到。


````python
def add(a, b):
    """两个数相加并返回结果。   ← 这是 docstring，必须是函数体第一行

    参数:
        a: 第一个数
        b: 第二个数
    返回:
        两数之和
    """
    # 这是注释：解释实现细节，解释器完全忽略，运行时访问不到
    return a + b


print(add.__doc__)   # ✅ 能打印出 docstring
help(add)            # ✅ 会显示完整的 docstring
````

## 6. 拷贝机制

### 6.1 别名、浅拷贝与深拷贝

- **别名**：`b = a`，b 和 a 指向同一个对象，改一个另一个也变。
- **浅拷贝**：`b = list(a)` / `a.copy()`，新建外层容器，但内层元素仍共享。
- **深拷贝**：`b = copy.deepcopy(a)`，递归复制所有层级，完全独立。


````python
# 1) 别名：b 和 a 指向同一个对象，改 b 等于改 a
a = ['1', '2', '3']
b = a                        # 别名（同一对象）
b.append('white')
print(b)                     # ['1', '2', '3', 'white']
print(a)                     # ['1', '2', '3', 'white']
print("-----------------")

# 2) 浅拷贝：新建外层列表，内层元素仍共享
a = ['1', '2', '3']
b = list(a)                  # 浅拷贝
b.append('white')
b[0] = '4'
print(b)                     # ['4', '2', '3', 'white']
print(a)                     # ['1', '2', '3']  ← 外层改动不影响 a

a = [[1, 2], [3, 4]]
b = list(a)                  # 浅拷贝
b[0].append(99)
print(b)                     # [[1, 2, 99], [3, 4]]
print(a)                     # [[1, 2, 99], [3, 4]]  ← 内层可变元素仍共享！
print("-----------------")

# 3) 深拷贝：递归复制所有层级，完全独立
import copy
b = copy.deepcopy(a)         # 深拷贝（新列表且新元素）
a.append('white')
print(a)                     # [[1, 2, 99], [3, 4], 'white']
print(b)                     # [[1, 2, 99], [3, 4]]  ← 完全不受影响
````

### 6.2 何时该用深拷贝

**一、为什么不能"有拷贝就 deepcopy"**

- 太慢、太占内存：深拷贝要递归遍历整个对象树，元素越多层越深越夸张。
- 对很多对象它处理不了或处理得很怪：
  - 文件句柄、数据库连接、网络 socket、线程锁——这些"资源型对象"不该被复制，deepcopy 要么报错要么复制出没用的幽灵对象。
  - 带循环引用的对象（A 引用 B，B 又引用 A），deepcopy 有专门处理，但逻辑复杂、容易踩坑。
  - 函数、类对象、模块——复制它们通常没意义。

**二、工程里的正确取舍顺序（从优到劣）**

`设计成不可变 > 浅拷贝 > 针对性构造新对象 > deepcopy`

- **首选不可变设计**：如果元素本身是 `int`/`str`/`tuple`/`frozenset` 这种不可变对象，那浅拷贝（`list(a)`、`a.copy()`）就够了——根本不存在"改元素内部"的可能。
- **浅拷贝能解决 90% 场景**：只复制外层，内层共享——只要你不原地改内层可变元素，就安全又高效。
- 实在要独立嵌套副本，再考虑 `deepcopy`。

**那 deepcopy 到底在哪些真实场景才上？**

需要"完全独立的嵌套可变副本"时才用，例如：

- **回溯 / 搜索算法（DFS）**：每一步要保存当前状态快照，提交/回退互不干扰。比如解数独、八皇后、全排列，经常 `deepcopy(board)` 一份拿去试。


## 7. 面向对象进阶

### 7.1 命名空间与作用域

Python 的变量查找遵循 LEGB 规则（Local → Enclosed → Global → Built-in）。`global` 和 `nonlocal` 关键字用来在函数内部显式修改外层变量；类体本身也是一个命名空间，类变量会被所有实例共享。


````python

a=1

def test1():
    global a
    a=2

test1()
print(a)


def test2():
    b=3
    def test3():
        nonlocal b
        b=4
    test3()
    print(b)

test2()

````

````python
class B:
    lst = []            # 类变量：所有实例共享同一个列表

b1 = B()
b1.lst = [2, 2]        # ⚠️ 赋值 → 给 b1 新建实例属性 lst，遮蔽了类变量
b2 = B()
b2.lst = [1, 1]        # 同理，b2 也有了自己的实例属性 lst
print(b1.lst)          # [2, 2]
print(b2.lst)          # [1, 1]
print(B.lst)           # []  ← 类变量没被改！赋值是新建实例属性，不是修改类变量
````

````python
class C:
    lst = []            # 类变量：所有实例共享同一个列表

c1 = C()
c1.lst.append(1)       # ✅ append → 沿查找链找到类变量 C.lst，原地修改它
print(c1.lst)          # [1]
print(C.lst)           # [1]  ← 类变量被改了！append 是原地修改，不是赋值
````

### 7.2 嵌套类

Python 允许在一个类的里面再定义一个类：

```python
class A:        # 外部类
    class B:    # 嵌套类
        pass
```

这**≠继承**！`B` 不是 `A` 的子类，它只是名字 `B` 被放进 `A` 的命名空间里。要访问它得写成 `A.B`。

**什么时候用？** 数据结构里非常常见：

- `LinkedList` 里面嵌套一个 `Node` 类
- `Tree` 里面嵌套一个 `Position` 类

这样做有两个好处：

1. **避免命名冲突**：`Node` 这个类名不会污染全局命名空间。
2. **表达"这个节点只属于这种数据结构"的语义**：链表节点天然属于链表。


### 7.3 `__slots__` —— 给实例属性"瘦身"

**1）默认情况：每个实例背后是个 dict**

Python 默认用 `__dict__` 存实例属性，比如：

```python
c = CreditCard(...)
c.__dict__      # {'_customer': ..., '_balance': ..., ...}
```

字典查找快，但本身要额外内存。如果你要创建成千上万个小对象（比如链表节点），这笔开销就明显了。

**2）`__slots__` 的作用**

`__slots__` 是一个类级元组，告诉 Python：这个类的实例只允许有这几个属性，不要用字典了：

```python
class CreditCard:
    __slots__ = '_customer', '_bank', '_account', '_balance', '_limit'
```

效果：

- 实例不再创建 `__dict__`，省内存。
- 你不能再给实例随便加新属性：`c.foo = 1` 会报 `AttributeError`。

**3）继承时的 `__slots__`**

如果父类声明了 `__slots__`，子类也必须声明，而且只写子类新增的属性：

```python
class PredatoryCreditCard(CreditCard):
    __slots__ = '_apr'    # 只补充新的；继承来的属性已由父类处理
```

> 书里的小提示：为了代码简洁，正文章节不一定每处都用 `__slots__`；但在节点类这种"实例超多"的地方会显式使用。


### 7.4 运算符重载

Python 的运算符在底层都对应一个魔法方法（如 `a + b` 实际调用 `a.__add__(b)`）。在自定义类中重写这些方法，就能改变运算符的行为。入门篇已演示过基础用法（`Matrix` 的 `__add__`），下面用 `Vector` 类做更完整的实现。


````python
class Vector:
    def __init__(self, d):
        self._coords = [0] * d          # 初始化 d 维，全是 0

    def __len__(self):
        return len(self._coords)        # len(v) 返回维数

    def __getitem__(self, j):
        return self._coords[j]          # v[j] 取值

    def __setitem__(self, j, val):
        self._coords[j] = val           # v[j] = val 赋值

    def __add__(self, other):
        if len(self) != len(other):     # 维度必须一致
            raise ValueError("dimension mismatch")
        result = Vector(len(self))
        for j in range(len(self)):
            result[j] = self[j] + other[j]   # 逐维相加
        return result

    def __eq__(self, other):
        return self._coords == other._coords # 列表逐元素比较

    def __ne__(self, other):
        return not self == other        # 不等于 = 取反等于

    def __repr__(self):
        return"《" + str(self._coords)[1:-1] + "》"
        
a=Vector(3)
b=Vector(3)
b[0]=1
b[1]=2
b[2]=3

# 这些新定义的对象实例，需要有自己的运算法则，所以需要重载运算符
print(a==b)
print(a+b)
````

### 8.2 用链表实现栈


> 💡 `def __len__(self)` 是 Python 规定的特殊接口，可以让对象支持内置语法 `len(obj)`。下面的 `LinkedStack` 就实现了它。


````python
# 用链表实现栈

class LinkedStack:

    class _Node:
        __slots__ = '_element', '_next'
        def __init__(self, element, next):
            self._element = element  # _element 表示外部最好不要动
            self._next = next

    def __init__(self):
        self._head = None
        self._size = 0

    def __len__(self):
        return self._size

    def is_empty(self):
        return self._size == 0

    def push(self, e):  # 入栈，相当于在链表头部插入
        self._head = self._Node(e, self._head)
        self._size += 1

    class MyListIsEmpty(Exception):  # 定义一种异常类型，用于表示栈为空
        pass

    def top(self):
        if self.is_empty():
            raise self.MyListIsEmpty('Stack is empty')  # 栈空时主动抛出异常
        return self._head._element

    def pop(self):
        if self.is_empty():
            raise self.MyListIsEmpty('Stack is empty')  # 栈空时主动抛出异常
        answer = self._head._element
        self._head = self._head._next
        self._size -= 1
        return answer


def main():
    s = LinkedStack()
    s.push("a")
    s.push("b")
    s.push("c")
    print(s.top())
    s.pop()
    print(s.top())
    s.pop()
    print(s.top())
    s.pop()

    try:
        print(s.top())  # 这里应该抛出异常
    except LinkedStack.MyListIsEmpty as e:
        print(f"错误类型：{type(e).__name__}")
        print(f"错误消息：{e}")
        print(f"错误位置：{e.__traceback__.tb_frame.f_code.co_filename}:{e.__traceback__.tb_lineno}")

if __name__ == '__main__':
    main()
````

### C-2.26　反向迭代器

书 2.3.4 节的 `SequenceIterator` 是前向迭代器。请实现一个 `ReversedSequenceIterator` 类，作为任意 Python 序列的反向迭代器：第一次 `next()` 返回序列最后一个元素，第二次返回倒数第二个，依此类推。


````python
class ReversedSequenceIterator:
    def __init__(self, sequence: list):
        self._seq = sequence
        self._cur = len(sequence) 
    
    def __next__(self):
        self._cur -= 1
        if self._cur == -1:
            raise StopIteration
        return self._seq[self._cur]

    def __iter__(self):
        return self

a = ReversedSequenceIterator([1, 2, 3])

for x in a:
    print(x)
````

### C-2.25　Vector 的 `__mul__` 双语义

给 `Vector` 实现 `__mul__`，让同一个 `*` 同时支持两种语义：

- **标量乘**：`v * 2` → 每个分量都乘 2（返回新 `Vector`）
- **向量点乘**：`u * v` → 对应分量乘积之和（返回一个数）

用 `isinstance` 在运行时区分这两种情况；另外要补 `__rmul__`，否则 `3 * v` 跑不通。


````python

class Vector:
    def __init__(self, arg1: list | int):
        try:
            self._values = [0]*arg1
            self._dims = arg1
        except TypeError:
            self._values = list(arg1)
            self._dims = len(arg1)
    
    def __getitem__(self, k): #  使得我可以使用 a[k], 而不需要使用 a.values[k]
        return self._values[k]
        
    def __setitem__(self, k, v): # 使得我可以使用 a[k] 
        self._values[k] = v

    def __len__(self):
        return self._dims
 
    def __sub__(self, other: Vector) -> Vector:
        if self._dims != other._dims:
            raise ValueError("Vector dimensions must be the same")
        result = Vector(len(self))        
        for i in range(len(self)):
            result[i] =self[i] - other[i]
        return result

    def __mul__(self, other: float | Vector) -> float|Vector:
        if isinstance(other, Vector):
            result=0
            for i in range(len(self)):
                result+=self[i]*other[i]
            return result
        elif isinstance(other, (float, int)):
            result=Vector(len(self))
            for i in range(len(self)):
                result[i]=self[i]*other
            return result
        else:
            raise TypeError("Vector must be multiplied by a int, float or a vector")

    def __rmul__(self, other: float | Vector) -> float|Vector:
        if  isinstance(other, (float, int)):
            result=Vector(len(self))
            for i in range(len(self)):
                result[i]=self[i]*other
            return result
        else:
            raise TypeError("Vector must be multiplied by a int, float or a vector")
        

    def __repr__(self):
        return f"Vector({self._values})"

a=Vector([1,2,3])
b=Vector([1,2,3])

print(a*b)

print(a*2)
print(2*a)


        
````

### 多边形继承体系（项目题）

基于拥有抽象方法 `area()` 和 `perimeter()` 的 `Polygon` 类发展继承层次结构。实现扩展自基类的 `Triangle`、`Quadrilateral`、`Pentagon`、`Hexagon` 和 `Octagon` 类，并给出有实际意义的 `area()` 和 `perimeter()` 方法。同时实现 `IsoscelesTriangle`、`EquilateralTriangle`、`Rectangle` 和 `Square` 类，让它们之间有合适的继承关系。

最后，写一个简单的程序：允许用户创建各种类型的多边形，输入几何尺寸，输出面积和周长。

> 🌟 **附加功能**：允许用户通过指定顶点坐标输入多边形，并能够测试两个多边形是否相似。


````python
from abc import ABC, abstractmethod
import math

class Polygon (ABC):
    def __init__(self):
        pass

    @abstractmethod
    def area(self) -> float:
        ...

    @abstractmethod
    def perimeter(self) -> float:
        ...

class Triangle(Polygon):
    def __init__(self,a,b,c:float):
        if a+b<=c or a+c<=b or b+c<=a:
            raise ValueError("三角形的边长不能构成三角形")
        self.a = a
        self.b = b
        self.c = c
        super().__init__()

    def area(self) -> float:
        p = (self.a + self.b + self.c) / 2
        return math.sqrt(p * (p - self.a) * (p - self.b) * (p - self.c))

    def perimeter(self) -> float:
        return self.a+self.b+self.c

class IsoscelesTriangle(Triangle):
    def __init__(self,a,b:float):
        super().__init__(a,b,b)

class EquilateralTriangle(IsoscelesTriangle):
    def __init__(self,a:float):
        super().__init__(a,a)


class Quadrilateral(Polygon):
    def __init__(self):
        pass

    @abstractmethod
    def area(self) -> float:
        ...
    @abstractmethod
    def perimeter(self) -> float:
        ...

class Rectangle(Quadrilateral):
    """
    矩形类, 允许输入两条边长或两个坐标点
    """
    def __init__(self,a,b:float):
        if isinstance(a,(float,int)) and isinstance(b,(float,int)):
            self.a = a
            self.b = b
        elif isinstance(a,tuple) and isinstance(b,tuple):
            self.a,self.b=  self.cal(a,b)
        else:
            raise ValueError("输入的参数类型错误")

        super().__init__()

    def cal(self,pot1,pot2):
            return (math.fabs(pot1[0]-pot2[0]),math.fabs(pot1[1]-pot2[1]))

    def area(self) -> float:
        return self.a*self.b

    def perimeter(self) -> float:
        return 2*(self.a+self.b)

class Square(Rectangle):
    """
    正方形类, 允许输入一条边长或两个坐标点
    """
    def __init__(self,a:float,b=None):
        if b is None:
            self.a = a
            self.b = a
            super().__init__(self.a,self.b)
        elif b is not None and isinstance(a,(tuple)) and isinstance(b,tuple):
            self.a,self.b=  super().cal(a,b)
            if self.a!=self.b:
                raise ValueError("正方形的边长必须相等")
            super().__init__(self.a,self.b)
        else:
            raise ValueError("输入的参数类型错误")


a=EquilateralTriangle(3)
print(a.area())
print(a.perimeter())

b=Square(3)
print(b.area())
print(b.perimeter())

c=Rectangle((3,0),(0,4))
print(c.area())
print(c.perimeter())

d=Square((0,0),(4,4))
print(d.area())
print(d.perimeter())
````

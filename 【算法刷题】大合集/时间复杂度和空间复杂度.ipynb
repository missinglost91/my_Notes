# 空间复杂度

```python
def funtion4(n):
    if n>0:
        funtion4(n-1)
        print(n)

funtion4(5)
```

```python

# 递归函数解决汉诺塔问题
def funtion5(n):
    if n>2:
        return 2*funtion5(n-1)+1
    else:
        return 3
    
print(funtion5(5))
```

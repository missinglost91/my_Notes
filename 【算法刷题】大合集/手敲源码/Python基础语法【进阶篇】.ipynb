## 错误类型与异常处理

### 常见运行时错误

计算机科学先驱 Grace Hopper: 请求原谅比得到许可更容易

try-except：只要在异常发生时，有一个异常处理的机制就行

| 错误类型 | 说明 | 示例 |
|---------|------|------|
| `NameError` | 使用了未定义的变量 | `print(x)`  # x 没定义 |
| `TypeError` | 类型操作错误 | `"hello" + 1` |
| `IndexError` | 索引超出范围 | `lst = [1,2]; lst[5]` |
| `KeyError` | 字典键不存在 | `d = {}; d['key']` |
| `ValueError` | 值错误 | `int("abc")` |
| `ZeroDivisionError` | 除以零 | `1 / 0` |
| `FileNotFoundError` | 文件不存在 | `open("no_file.txt")` |
| `AttributeError` | 属性不存在 | `None.some_method()` |

````python


try: #只保护可能出错的代码
    num = int(input("请输入数字："))
    result = 10 / num
    
    
except ValueError as e:
    # 2. 捕获类型错误
    print(f"错误类型：{e}")
    print("输入的不是数字!")
    
except ZeroDivisionError as e:
    # 3. 捕获除零错误
    print(f"错误类型：{e}")
    print("不能除以零！")

else:    
    print(f"结果是：{result}")

    
finally:
    # 6. 无论是否异常都执行（常用于清理资源）
    print("程序结束")

````

### 1️ 可迭代对象 (Iterable)
- 定义 ：能被 for 循环遍历的对象
- 判断方法 ：实现了 __iter__() 方法 或 __getitem__() 方法
- 常见例子 ： list , tuple , dict , set , str

````python
from collections.abc import Iterable

lst = [1, 2, 3]
it = iter(lst) 

# 可以被 for 循环
print("--------迭代对象第一次遍历---------")
for i in lst:
    print(i)

print("--------迭代对象第一次遍历---------")
for i in lst:
    print(i)

print("--------迭代器第一次遍历---------")
for i in it: 
    print(i)

print("--------迭代器第二次遍历---------")
for i in it:
    print(i) #不支持第二遍，没有元素了

print("----------迭代器取next---------")
it = iter(lst) 
print(next(it))  # 1
print(next(it))  # 2
print(next(it))  # 3
````

## 序列类型的打包和解包

### 打包 (Packing)
把多个值组合成一个序列

```python
# 隐式打包
t = 1, 2, 3
print(t)  # (1, 2, 3)
```

### 解包 (Unpacking)
把序列拆分成多个独立的值

| 类型 | 语法 | 说明 | 示例 |
|------|------|------|------|
| **基础解包** | `a, b, c = seq` | 变量数必须等于元素数 | `a, b = [1, 2]` |
| **星号解包** | `a, *rest = seq` | `*` 收集剩余元素为列表 | `a, *rest = [1, 2, 3]` |
| **只取部分** | `a, *_ = seq` | `_` 表示丢弃 | `first, *_ = [1, 2, 3]` |
| **函数参数** | `func(*args)` | 列表解包为位置参数 | `add(*[1, 2])` |
| **字典参数** | `func(**kwargs)` | 字典解包为关键字参数 | `func(**{'a': 1})` |

### 示例代码

```python
# 收集中间部分
first, *middle, last = [1, 2, 3, 4, 5]
print(first)    # 1
print(middle)   # [2, 3, 4]  ← 自动打包成列表
print(last)     # 5

# 函数多返回值解包
def get_user_info():
    return "Alice", 25, "Engineer"

name, age, job = get_user_info()
print(name)  # Alice
print(age)   # 25
print(job)   # Engineer
```

### 常见错误

```python
#  错误 1：数量不匹配
a, b = [1, 2, 3]  # ValueError: too many values to unpack

# ✅ 正确：使用星号
a, b, *rest = [1, 2, 3]

# ❌ 错误 2：多个星号
*a, *b = [1, 2, 3, 4]  # SyntaxError: two starred expressions

# ✅ 正确：只能有一个星号
a, *b, c = [1, 2, 3, 4]
```

## 栈和队列

### 栈 (Stack)

我们用 `S` 来表示栈这一 **ADT(抽象数据类型)**

**特点**：后进先出 (LIFO - Last In First Out)

#### 栈的基本操作

| 操作 | 方法 | 说明 | 时间复杂度 |
|------|------|------|----------|
| 入栈 | `S.push(e)` | 将元素 e 压入栈顶 | O(1) |
| 出栈 | `S.pop()` | 弹出并返回栈顶元素 | O(1) |
| 查看栈顶 | `S.top()` | 返回栈顶元素（不弹出） | O(1) |
| 判空 | `S.is_empty()` | 判断栈是否为空 | O(1) |
| 长度 | `len(S)` | 返回栈中元素个数 (通过 `__len__` 实现) | O(1) |

```python
# 示例用法
S.push(e)      # 入栈
S.pop()        # 出栈
S.top()        # 查看栈顶
S.is_empty()   # 判空
len(S)         # 获取长度
```

# 链表


def __len__(self)
是 Python 规定的特殊接口，可以让对象支持内置语法 len(obj)。

````python
# 用链表实现栈

class LinkedStack:

    class _Node:
        __slots__ = '_element', '_next'
        def __init__(self, element,next):
            self._element=element  #_element 表示外部最好不要动
            self._next = next
    

    def __init__(self):
        self._head = None
        self._size = 0
    
    def __len__(self):
        return self._size
    
    def is_empty(self):
        return self._size == 0
    
    def push(self,e): # 入栈,想当年与链表的头
        self._head = self._Node(e,self._head)
        self._size += 1
    
    class MyListIsEmpty(Exception): # 定义一种异常类型，用于表示栈为空
        pass

    def top(self):
        if self.is_empty():
            raise self.MyListIsEmpty('Stack is empty') #主动抛出异常
        return self._head._element

    def pop(self):
        if self.is_empty():
            raise self.MyListIsEmpty('Stack is empty') #主动抛出异常
        answer = self._head._element
        self._head = self._head._next
        self._size -= 1
        return answer


def main():
    s = LinkedStack()
    s.push("a")
    s.push("b")
    s.push("c")
    print(s.top())
    s.pop()
    print(s.top())
    s.pop()
    print(s.top()) 
    s.pop()

    try:
        print(s.top())  #这里应该抛出异常 
    except LinkedStack.MyListIsEmpty as e:
        print(f"错误类型：{type(e).__name__}")
        print(f"错误消息：{e}")
        print(f"错误位置：{e.__traceback__.tb_frame.f_code.co_filename}:{e.__traceback__.tb_lineno}")

if __name__ == '__main__':
    main()
````

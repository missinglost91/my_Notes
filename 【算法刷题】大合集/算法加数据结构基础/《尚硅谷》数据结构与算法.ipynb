# 数据结构与算法


```python
def sum(nums):
    total = 0
    for num in nums:
        total += num
    return total
```

```python
#冒泡排序 O(n^2)
def bubble_sort(arr):
    for i in range(len(arr)):
        for j in range(len(arr)-1-i):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]



            
```

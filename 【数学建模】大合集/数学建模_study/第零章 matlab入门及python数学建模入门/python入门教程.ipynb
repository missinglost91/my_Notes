## numpy库

```python
import numpy as np  #引入numpy库
```

```python
arr=np.array([[1,3,5,4],  #用np.array创建一个数组对象
               [1,2,4,4],
               [3,4,5,7]])
```

```python
print(arr) #打印数组
```

```python
print(arr.shape) #判断数组各维度大小，三行四列
```

```python
arr[0,:] #取第一行
```

```python
arr[:,0] #取第一列
```

```python
arr[0][0]  #取第一行第一列的数
```

```python
#直接加减乘除是对应位置上加减乘除
a1 = np.array([1,2,3,4,5])
a2 = np.array([1,2,3,4,5])

print("a1+a2=",a1+a2)
print("a1*a2=",a1*a2)
print("a1/a2=",a1/a2)
```

```python
# reshape方法
new_arr=arr.reshape(2,6) #元素个数应该保持不变
print(new_arr)
print(new_arr.shape)
```

```python
# 转置矩阵
ts_arr=arr.transpose()
print(ts_arr)
```

```python
#NumPy 统计函数
# 1是按行统计，0是按列统计
```

```python
arr=np.array([[1,3,5,4],  #用np.array创建一个数组对象
               [1,2,4,4],
               [3,4,5,7]])
```

```python
np.amin(arr,1) #按行统计最小值
```

```python
# np.argsort(x) 返回索引排序后的结果

x = np.array([3,  1,  2])
# 3是索引0，1是索引1，2是索引2
# 从小到大排序后 ，数组变成1，2，3
# 即是索引1，索引2，索引0
y = np.argsort(x)
print (y)
```

```python
#numpy.extract() 函数根据某个条件从数组中抽取元素，返回满条件的元素

x = np.arange(9.).reshape(3,  3)
print ('我们的数组是：')
print (x)
# 定义条件, 选择偶数元素
condition = np.mod(x,2)  ==  0
print ('按元素的条件值：')
print (condition)
print ('使用条件提取元素：')
print (np.extract(condition, x))

```

数据的保存和导入 .npy后缀名

```python
a = np.array([[10, 9, 4],
              [3, 4, 1]])
np.save('a.npy', a)
np.save('b.npy', a)
np.save('c.npy', a)

arr=np.load('a.npy')
print(arr)
```

```python
# 设置随机种子，保证结果可重复
np.random.seed(42)

# 生成[0,1)均匀分布的2x3随机数组
print(np.random.rand(2, 3))

# 生成1到10之间的随机整数（不含10），形状为(3,4)
print(np.random.randint(1, 10, size=(3, 4)))

# 生成均值0、标准差1的3x3正态分布随机数组
print(np.random.normal(0, 1, size=(3, 3)))



arr=np.random.randint(1, 20, size=(4, 4))

print(np.sum(arr[arr <= 10]))
```

## 核函数
核函数注意事项
1. 核函数只能访问GPU内存
2. 核函数不能使用变长参数
3. 核函数不能使用静态变量
4. 核函数不能使用函数指针
5. 核函数具有异步性


```cpp
int main(void)
{
    主机代码
    核函数调用
    主机代码
    return 0;
}
```
核函数不支持C++的iostream*

```python
import cv2   #opencv读取的格式是BGR
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline





```

```python

def cv_show(name,img):
    cv2.imshow('pig',img)  #image弹出的图片展示窗口名称，img为对象名
    cv2.waitKey(0)   #按任意键退出
    cv2.destroyAllWindows()
```

```python
img=cv2.imread('pig.jpg')

pig=img[50:200,200:400]  #前一个区域代表纵区域，后面的代表横区域

#cv_show('eye',pig)
```

颜色通道提取

```python
b,g,r=cv2.split(img)
```

```python
#只保留R
cur_img=img.copy()
cur_img[:,:,0]=0 #所有像素点的b通道为0
cur_img[:,:,1]=0 #所有像素点的g通道为0
cv_show('R',cur_img)
```

```python
#只保留G
cur_img=img.copy()
cur_img[:,:,0]=0 #所有像素点的b通道为0
cur_img[:,:,2]=0 #所有像素点的g通道为0
cv_show('G',cur_img)
```

```python
#只保留B
cur_img=img.copy()
cur_img[:,:,1]=0 #所有像素点的b通道为0
cur_img[:,:,2]=0 #所有像素点的g通道为0
cv_show('B',cur_img)
```

```python
img=cv2.merge([b,g,r])
cv_show('again-img',img)
```

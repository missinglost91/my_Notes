```python
import cv2  #opencv读取的格式是BGR
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline

def cv_show(name, img):
    cv2.imshow('pig', img)  #image弹出的图片展示窗口名称，img为对象名
    cv2.waitKey(0)  #按任意键退出
    cv2.destroyAllWindows()

img_cat=cv2.imread('cat.jpg')
img_dog=cv2.imread('dog.jpg')
```

```python
img_cat2=img_cat+10
img_cat[:5,:,0] #取前五行，全部列，r通道的数值
```

```python
# 如果你的img_cat是用cv2.imread()读取的，OpenCV 默认会把图像存为BGR 通道顺序，而 matplotlib 需要 RGB，所以需要转换通道
# img_cat_rgb = cv2.cvtColor(img_cat, cv2.COLOR_BGR2RGB)
```

```python
img_cat.shape
```

```python
img_dog=cv2.resize(img_dog,(1944,1458))
```

```python
img_add=cv2.add(img_cat,img_dog) #直接融合
img_addweighted=cv2.addWeighted(img_cat,0.5,img_dog,0.5,0) #权重融合

plt.subplot(221),plt.imshow(cv2.cvtColor(img_cat, cv2.COLOR_BGR2RGB)),plt.title('cat')
plt.subplot(222),plt.imshow(cv2.cvtColor(img_dog, cv2.COLOR_BGR2RGB)),plt.title('dog')

plt.subplot(223),plt.imshow(cv2.cvtColor(img_add, cv2.COLOR_BGR2RGB),'gray'),plt.title('add')
plt.subplot(224),plt.imshow( cv2.cvtColor(img_addweighted, cv2.COLOR_BGR2RGB),'gray'),plt.title('addweighted')
plt.tight_layout() #自动优化子图间的距离
```

```python
import cv2   #opencv读取的格式是BGR
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline

vc=cv2.VideoCapture('maomao.mp4')



```

```python
if vc.isOpened():
    open,frame=vc.read() #用open来判断视频是否正常打开
else:
    open=False
```

```python
while open:
    ret,frame=vc.read() #用ret来判断视频每一帧是否正常打开
    if frame is None:
        break
    if ret== True:
        gray =cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
        cv2.imshow('result',gray)
        if cv2.waitKey(10) & 0xFF == ord('q'): #退出键'q'
            break
vc.release()
cv2.destroyAllWindows()
```

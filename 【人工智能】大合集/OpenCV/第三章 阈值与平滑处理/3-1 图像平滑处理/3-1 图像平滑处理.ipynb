```python
import cv2  #opencv读取的格式是BGR
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline

def cv_show(name, img):
    cv2.imshow(name, img)  #image弹出的图片展示窗口名称，img为对象名
    cv2.waitKey(0)  #按任意键退出
    cv2.destroyAllWindows()

img_kun=cv2.imread('kun_zao.jpg')

```

```python
#均值滤波
blur=cv2.blur(img_kun,(4,4))
```

```python
#方框滤波
box=cv2.boxFilter(img_kun,-1,(5,5),normalize=True) #-1是通道数一致,true会除25，其实和均值滤波一样
```

```python
#方框滤波
box2=cv2.boxFilter(img_kun,-1,(3,3),normalize=False) #-1是通道数一致,False不会除25
```

```python
#高斯滤波
#卷积满足高斯分布，更重视中间数据
assian=cv2.GaussianBlur(img_kun,(5,5),1)
```

```python
#中值滤波
median=cv2.medianBlur(img_kun,5)
```

```python
res=np.hstack((blur,assian,median))
cv_show('res',res)
```

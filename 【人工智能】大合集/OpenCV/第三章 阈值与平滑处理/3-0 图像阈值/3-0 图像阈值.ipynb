```python
import cv2  #opencv读取的格式是BGR
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline

def cv_show(name, img):
    cv2.imshow('pig', img)  #image弹出的图片展示窗口名称，img为对象名
    cv2.waitKey(0)  #按任意键退出
    cv2.destroyAllWindows()

img_cat=cv2.imread('cat.jpg', cv2.IMREAD_GRAYSCALE)

img_dog=cv2.imread('dog.jpg')


```

```python
ret,img_cat_BINARY=cv2.threshold(img_cat,127,255,cv2.THRESH_BINARY)
ret,img_cat_BINARY_INV=cv2.threshold(img_cat,127,255,cv2.THRESH_BINARY_INV)
ret,img_cat_TRUNC=cv2.threshold(img_cat,127,255,cv2.THRESH_TRUNC)
ret,img_cat_TOZERO=cv2.threshold(img_cat,127,255,cv2.THRESH_TOZERO)
ret,img_cat_TOZERO_INV=cv2.threshold(img_cat,127,255,cv2.THRESH_TOZERO_INV)
```

```python
plt.subplot(231),plt.imshow(img_cat,'gray'),plt.title('origin')
plt.subplot(232),plt.imshow(img_cat_BINARY,'gray'),plt.title('BINARY')

plt.subplot(233),plt.imshow( img_cat_BINARY_INV,'gray'),plt.title('BINARY_INV')
plt.subplot(234),plt.imshow( img_cat_TRUNC,'gray'),plt.title('TRUNC')
plt.subplot(235),plt.imshow( img_cat_TOZERO,'gray'),plt.title('TOZERO')
plt.subplot(236),plt.imshow( img_cat_TOZERO_INV,'gray'),plt.title('TOZERO_INV')
plt.tight_layout() #自动优化子图间的距离
```

我们可以用数组来优化一下简化写法

```python
ret,img_cat_BINARY=cv2.threshold(img_cat,127,255,cv2.THRESH_BINARY)
ret,img_cat_BINARY_INV=cv2.threshold(img_cat,127,255,cv2.THRESH_BINARY_INV)
ret,img_cat_TRUNC=cv2.threshold(img_cat,127,255,cv2.THRESH_TRUNC)
ret,img_cat_TOZERO=cv2.threshold(img_cat,127,255,cv2.THRESH_TOZERO)
ret,img_cat_TOZERO_INV=cv2.threshold(img_cat,127,255,cv2.THRESH_TOZERO_INV)

title=['origin','THRESH_BINARY','THRESH_BINARY_INV','THRESH_TRUNC','THRESH_TOZERO','THRESH_TOZERO_INV']
img=[img_cat,img_cat_BINARY,img_cat_BINARY_INV,img_cat_TRUNC,img_cat_TOZERO,img_cat_TOZERO_INV]
for i in range(6):
    plt.subplot(2,3,i+1), plt.imshow(img[i], 'gray'), plt.title(title[i]), plt.xticks([]), plt.yticks([])
plt.tight_layout()  #自动优化子图间的距离
```

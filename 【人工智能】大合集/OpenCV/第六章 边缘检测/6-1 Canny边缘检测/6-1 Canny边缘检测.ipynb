```python
import cv2  #opencv读取的格式是BGR
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.image import imread
%matplotlib inline

def cv_show(name, img):
    cv2.imshow(name, img)  #image弹出的图片展示窗口名称，img为对象名
    cv2.waitKey(0)  #按任意键退出
    cv2.destroyAllWindows()

img_mount=cv2.imread('mount.jpg', cv2.IMREAD_GRAYSCALE)
img_lena=cv2.imread('lena.jpg', cv2.IMREAD_GRAYSCALE)

v1=cv2.Canny(img_lena, 80, 150)#下限和上限
v2=cv2.Canny(img_lena, 50, 100)
```

```python
res=np.hstack((v1,v2))
cv_show('lena.jpg', res)
```

```python
import cv2  #opencv读取的格式是BGR
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.image import imread
%matplotlib inline

def cv_show(name, img):
    cv2.imshow(name, img)  #image弹出的图片展示窗口名称，img为对象名
    cv2.waitKey(0)  #按任意键退出
    cv2.destroyAllWindows()

img = cv2.imread('lena.jpg')
template = cv2.imread('lena_face.jpg')

plt.imshow(img[:,:,::-1])#颜色通道翻转

```

```python

plt.imshow(template[:,:,::-1])#颜色通道翻转
```

```python
h,w= template.shape[:2] # 获取模板图像的尺寸
print(h,w)
```

```python
res = cv2.matchTemplate(img, template, cv2.TM_CCOEFF_NORMED) # 该函数返回一个结果矩阵，其中每个元素表示目标图像中对应位置与模板图像的相似度。最大值是最匹配的点

```

```python
min_val, max_val, min_loc, max_loc = cv2.minMaxLoc(res)


top_left = max_loc

bottom_right = (top_left[0]+w, top_left[1]+h)
```

```python

cv2.rectangle(img, top_left, bottom_right , (0, 255, 0), 2)

# 显示结果图像
plt.imshow(img[:,:,::-1])
```

# 节点篇

在 ROS2 中，节点（Node）是最小的计算单元。当我们在 Python 中使用 `class MyNode(Node):` 时，就是在使用面向对象编程中的“继承”机制，拿过 ROS2 官方提供的通用“员工说明书”（即 `rclpy.node.Node` 基类），从而让我们的自定义程序瞬间具备 ROS2 的核心能力。

这个父类（`Node`）主要提供了以下几大类的属性和方法：

 1. 初始化与基础属性
*   **节点名称（工牌）**：通过调用父类的初始化方法 `super().__init__('node_name')`，为节点分配一个全网唯一的标识符（相当于员工的工牌）。
*   **日志系统**：提供 `self.get_logger()` 方法。相比于普通的 `print` 语句，它支持不同级别的日志输出（如 `.info()`, `.warn()`, `.error()`），是排查问题的标准工具。
*   **参数管理**：允许你在运行时动态声明、获取和修改节点的运行参数，而无需重新编译代码。

1. 核心通信机制（话题 Pub/Sub）
这是节点最常用的功能，用于实现数据的单向流动：
*   **创建发布者**：`self.create_publisher(消息类型, '话题名', 队列大小)`。就像建立一个广播站，定期向指定频道发送数据。
*   **创建订阅者**：`self.create_subscription(消息类型, '话题名', 回调函数, 队列大小)`。就像调频收音机，一旦监听到特定频道的数据，就会自动触发你设定的“回调函数”进行处理。

1. 时间与服务控制
*   **定时器（Timer）**：`self.create_timer(周期秒数, 回调函数)`。类似于设置一个闹钟，让节点能够按固定的时间间隔周期性执行某项任务（例如每秒发布一次传感器状态）。
*   **服务端/客户端（Service）**：
    *   `self.create_service(...)`：提供服务，处理其他节点的请求并返回响应（类似餐厅服务员接单上菜）。
    *   `self.create_client(...)`：作为客户端发送请求，等待服务端的响应结果。

💡 关键注意事项（Python 的 `self` 陷阱）
当你继承了 `Node` 父类并在自己的类中重写了 `__init__` 方法时，**必须**在第一行显式调用 `super().__init__('节点名')`。如果不写这一行，你的子类就只是一具没有灵魂的躯壳，无法完成 ROS2 底层的初始化工作，后续调用 `create_publisher` 或 `get_logger` 等方法时就会报错。

## 2-1 编写第一个核心节点

```python
import rclpy
from rclpy.node import Node

def main():
    rclpy.init() #初始化工作
    node = Node('ros2_python_node') #node对象
    node.get_logger().info('Hello World!')
    node.get_logger().warn('Hello World!')
    node.get_logger().error('NoHello World!')
    rclpy.spin(node)
    rclpy.shutdown()

if __name__ == '__main__':
    # 主函数
    main()

   
```
